```python
import pandas as pd

input_file = "../Dataset/QK-article.csv"  # Đường dẫn đầy đủ trên máy bạn
output_file = "../Dataset/QK-article_subset_5M.csv"

target_rows = 5_000_000
chunksize = 500_000

subset_chunks = []
total_rows = 0

for chunk in pd.read_csv(input_file, chunksize=chunksize):
    if total_rows + len(chunk) >= target_rows:
        remaining = target_rows - total_rows
        subset_chunks.append(chunk.iloc[:remaining])
        break
    else:
        subset_chunks.append(chunk)
        total_rows += len(chunk)

subset_df = pd.concat(subset_chunks, ignore_index=True)
subset_df.to_csv(output_file, index=False)

print(f"✅ Đã lưu {output_file} với {subset_df.shape[0]} dòng.")
```

```python
import pandas as pd

# 📥 Đọc lại dataset đã tiền xử lý đầy đủ 5 triệu dòng
file_path = '../Dataset/QK-article_subset_5M_preprocessed.csv'
df = pd.read_csv(file_path)

# 🧠 Tính toán tỉ lệ mất cân bằng để dùng cho training sau
count_1 = df['read'].sum()
count_0 = len(df) - count_1
scale_pos_weight = round(count_0 / count_1, 4)

print(f"📉 Số lượng read = 1: {count_1}")
print(f"📉 Số lượng read = 0: {count_0}")
print(f"⚖️ Tỉ lệ scale_pos_weight (class 0 / class 1): {scale_pos_weight}")

# 💾 Ghi lại (nếu muốn đảm bảo tính thống nhất, ta chỉ cần ghi lại để giữ đúng định dạng kiểu dữ liệu)
df.to_csv(file_path, index=False)
print(f"✅ Đã ghi lại file giữ nguyên toàn bộ dữ liệu tại: {file_path}")
```

```python
import pandas as pd

# --- Video dataset ---
video_df = pd.read_csv("../Dataset/QK-video_subset_5M.csv")   # hoặc file chứa dữ liệu gốc
video_items = video_df["item_id"].unique().tolist()

# --- Article dataset ---
article_df = pd.read_csv("../Dataset/QK-article_subset_5M_preprocessed.csv")
article_items = article_df["item_id"].unique().tolist()

# Save lại để backend load
import json
with open("Deep FM/video/video_items.json", "w") as f:
    json.dump(video_items, f)

with open("Deep FM/article/article_items.json", "w") as f:
    json.dump(article_items, f)
```

```python
import pandas as pd
import json

# Giả sử bạn đã load DataFrame
# video_df = pd.read_csv("video.csv")
# article_df = pd.read_csv("article.csv")

# ✅ 1. Làm sạch dữ liệu cho VIDEO
# Thay "\\N" hoặc NaN bằng 0
video_df["video_category"] = video_df["video_category"].replace("\\N", 0).fillna(0)
# Ép kiểu int để chắc chắn dữ liệu chuẩn
video_df["video_category"] = video_df["video_category"].astype(int)

# ✅ 2. Làm sạch dữ liệu cho ARTICLE
# Thay "\\N" hoặc NaN bằng 0 cho cả 2 cột
article_df["category_first"] = article_df["category_first"].replace("\\N", 0).fillna(0)
article_df["category_second"] = article_df["category_second"].replace("\\N", 0).fillna(0)
# Ép kiểu int
article_df["category_first"] = article_df["category_first"].astype(int)
article_df["category_second"] = article_df["category_second"].astype(int)

# ✅ 3. Tạo mapping đã sạch
video_map = video_df.groupby("item_id")["video_category"].first().to_dict()
article_map = article_df.groupby("item_id")[["category_first", "category_second"]].first().to_dict("index")

# ✅ 4. Ghi đè lên file JSON cũ
with open("Deep FM/video/video_item_map.json", "w") as f:
    json.dump(video_map, f)

with open("Deep FM/article/article_item_map.json", "w") as f:
    json.dump(article_map, f)

print("✅ Đã làm sạch dữ liệu và ghi đè 2 file JSON:")
print("   - Deep FM/video/video_item_map.json")
print("   - Deep FM/article/article_item_map.json")
```
